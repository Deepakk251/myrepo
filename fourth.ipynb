```python
import nltk
import string 

s = ["The cat sat on the mat", "The dog sat on the log", "The fish is in the tank "]

corpus = " ".join(s)
print(corpus)
```

```python
#from nltk
from nltk.tokenize import word_tokenize,sent_tokenize

t = word_tokenize(corpus)
print(t)
```

```python
#removing punctuation

t = [w for w in t if w not in string.punctuation]
print(t)
```

```python
vocabuolary = set(t)
print(len(vocabuolary))
```

```python
import nltk
import string

nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python

s = ["I Love NLP","I Love Engineering","NLP is fun"]
corpus = " ".join(s)
print(corpus)
corpus = corpus.lower()
print(corpus)
token = word_tokenize(corpus)
print(token)

token = [w for w in token if w not in string.punctuation]
print(token)

vocab = set(token)
print(vocab)
print(len(vocab))
```

```python
word_index = {word: i for i , word in enumerate(vocab)}
print(word_index)
```

```python
def one_hot(word):
    vec = [0]*len(vocab)
    vec[word_index[word]]=1
    return vec

for i in word_index:
    print(i , one_hot("nlp"))
```
