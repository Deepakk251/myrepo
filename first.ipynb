```python
import nltk
nltk.download('brown')
from nltk.corpus import brown
```

```python
print(brown.categories())
```

```python
print(brown.words())
```

```python
print(brown.words(categories='news'))
```

```python
c = brown.categories()
for ci in c:
    w_c = len(brown.words(categories=ci))
    print(w_c)
```

```python
s_w = brown.words(categories='news')
v = set(s_w)
print(v)
```
